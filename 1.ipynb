```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch", "Headphones"]
sales = [500, 800, 300, 450, 600]
plt.bar(products, sales)
plt.xlabel("|Products|")
plt.ylabel("sales")
plt.title("Sales Market Weekly sales")

plt.grid()
plt.savefig("jhgf")

plt.show()
```

```python
products = ["Laptop", "Mobile", "Tablet", "Watch", "Headphones"]
sales = [500, 800, 300, 450, 600]
plt.barh(products, sales)
plt.xlabel("|Products|")
plt.ylabel("sales")
plt.title("Sales Market Weekly sales")

plt.grid()
plt.savefig("jhf")

plt.show()
```

```python
import numpy as np
# import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch", "Headphones"]
sales_2025 = [500, 800, 300, 450, 600]
sales_2026 = [650, 900, 420, 550, 750]

x = np.arange(len(products))
width = 0.35

plt.bar(x - width/2, sales_2025, width, label="2025")
plt.bar(x + width/2, sales_2026, width, label="2026")

plt.xticks(x, products)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Sales Comparison")
plt.legend()
plt.savefig("jhg")


plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch", "Headphones"]
sales = [500, 800, 300, 450, 600]

bars = plt.bar(products, sales)

plt.bar_label(bars, padding=3)

plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product Sales with Values")
plt.savefig("oiu")

plt.show()
```
